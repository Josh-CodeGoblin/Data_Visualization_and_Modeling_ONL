## The reproducible analysis workflow

1. **Question:** What do we want to learn?
2. **Data:** Which observations and features can help answer it?
3. **Evidence:** What should we calculate and display to answer the question?
4. **Check:** Does the result make sense in context? Do a few original records support the meaning of the calculation?
5. **Conclusion:** What claim is supported?
6. **Limitation:** What should we avoid concluding?

**Which observations and features will you use? Explain relevant units, any filtering, and how you will handle missing or not-applicable values using the dictionary. For a percentage, define the denominator.**

```python
# Prepare, filter, or calculate features as needed; display useful intermediate results.
# Calculate the percentage of orders in each order status.
order_df["OrderStatus"].value_counts(normalize=True) * 100
```

```python
# Calculate and display the evidence that answers your question. Add cells as needed.
```

**Which specific value or comparison answers your question? Explain how your calculations connect the selected data to that result.**

Your answer:
Pending orders made up the largest percentage of the dataset at 26.9%, followed closely by shipped orders at 26.8%. Processing orders represented 23.5%, and delivered orders represented 22.8%. I calculated these percentages by counting each OrderStatus, dividing by the total number of orders, and multiplying the proportions by 100.

### Step 4: Check

Perform a sanity check tied to the meaning of your question. Explain what values or patterns would make you suspicious and whether your result makes sense. Depending on the analysis, consider plausible values and units, whether indicators match original labels, or whether the ranges within comparison groups match their definitions.
```python
# Optional: add a targeted sanity check if code would help.
(order_df["OrderStatus"].value_counts(normalize=True) * 100).sum()
```

**What would make you suspicious, and why? Does your evidence make sense in context? Describe any issue you found and how you addressed it.**

Your answer:
I would be suspicious if the percentages did not add up to approximately 100% or if an unexpected order-status category appeared. The calculated percentages sum to 100%, so all orders are represented within the four status groups. The results therefore make sense for this comparison, and I did not identify an issue that required changing the calculation.

### Step 5: Conclusion


**Answer your research question in a sentence using the numerical evidence. How does it compare with your prediction?**

Your answer:
Pending orders represented the largest percentage of the dataset at 26.9%, followed by shipped orders at 26.8%, processing orders at 23.5%, and delivered orders at 22.8%. This did not match my prediction because I expected delivered orders to make up the largest percentage.

### Step 6: Limitation


**What should a reader avoid concluding from this result, and why? Connect your limitation to this question and dataset.**

Your answer:
These percentages describe only the distribution of order statuses in this dataset and should not be assumed to represent retail orders in general. The result also does not explain why an order has a particular status or whether the status changes over time.
